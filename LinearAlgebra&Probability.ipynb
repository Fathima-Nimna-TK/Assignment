LINEAR ALGEBRA AND PROBABILITY

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Task 1


```python
df = pd.read_csv("winequality-red.csv")

print("dataset shape (rows,columns): ",df.shape)
print("\nMissing values before replacement: ")
print(df.isnull().sum())

df = df.fillna(df.mean(numeric_only=True))

print("\nTotal missing values after replacement: ",df.isnull().sum().sum())
df.head()
```

Task 2

```python
alcohol = df['alcohol'].to_numpy()
citric_acid = df['citric acid'].to_numpy()

print('Alcohol vector:',alcohol[:5])
print('Citric acid vector:',citric_acid[:5])

print('Alcohol vector shape:',alcohol.shape)
print('Citric acid vector shape:',citric_acid.shape)
```

Task 3

```python
X = df[['alcohol','density']].to_numpy()
cov_matrix = np.cov(X.T)

print('Covariance metrix:')
print(cov_matrix)
```

Task 4


```python
eigenvalues, eigenvectors = np.linalg.eig(cov_matrix)
order = np.argsort(eigenvalues)[::-1]
eigenvalues = eigenvalues[order]
eigenvectors = eigenvectors[:, order]

print("Eigenvalues (Ascending):")
print(eigenvalues)
print("\nCorresponding eigenvectors (each column matches the eigenvalue above):")
print(eigenvectors)
```

Task 5

```python
quality_counts = df["quality"].value_counts().sort_index()

print("Wine quality distribution (count):")
print(quality_counts)

most_common_quality = int(df["quality"].mode()[0])
most_common_count = int(quality_counts.loc[most_common_quality])
most_common_percent = most_common_count / len(df) * 100

print("\nMost common quality score:", most_common_quality)
print("Number of wines with this score:", most_common_count)
print(f"Percentage of all wines: {most_common_percent:.2f}%")

quality_counts.plot(kind="bar", rot=0)
plt.title("Distribution of Wine Quality Scores")
plt.xlabel("Quality score")
plt.ylabel("Number of wines")
plt.tight_layout()
plt.show()
```

```python
**Interpretation:** Quality score 5 is the most common, with 681 wines (about 42.59%). Score 6 follows with 638 wines (about 39.90%). Together, scores 5 and 6 represent about 82.49% of the dataset. Scores 3 and 8 are uncommon, and the observed scores range from 3 to 8.
```
