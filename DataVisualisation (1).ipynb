```python
import seaborn as sns
sns.set(style="darkgrid")
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import warnings 
warnings.filterwarnings("ignore")
plt.rcParams["figure.figsize"]=(10,10)
```

```python
data=pd.read_csv("Titanic-Dataset.csv")
data.head()
```

1-COUNTPLOT

```python
plt.figure(figsize=(10,6))
sns.countplot(data=data,x='Embarked',hue='Sex')
plt.title('Passenger Count by Embarked Port and Sex')
plt.xlabel('Embarked port')
plt.ylabel('Number of Passenger')
plt.legend(title='Sex')
plt.show()
```

2-SWARMPLOT

```python
plt.figure(figsize=(10,6))
sns.swarmplot(data=data,x='Pclass',y='Age',hue='Sex',size=3)
plt.title('Age Distribution across Passenger Classes')
plt.xlabel('Passenger class')
plt.ylabel('Age')
plt.legend(title='Sex')
plt.show()
```

3-PIE CHART

```python
gender_count = data['Sex'].value_counts()
plt.figure(figsize=(7,7))
plt.pie(gender_count,labels=gender_count.index,autopct='%1.1f%%',explode=(0,0.1),startangle=90)
plt.title('Gender Distribution of Titanic Passengers')
plt.show()
```

4-HEAT MAP

```python
plt.figure(figsize=(10,6))
columns = ['Age','Fare','Pclass','SibSp','Parch','Survived']
correlation_matrix = data[columns].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Matrix of Titanic Dataset')
plt.show()
          
```

6-VIOLIN PLOT

```python
plt.figure(figsize=(10,6))
sns.violinplot(data=data, x='Pclass', y='Fare', hue='Sex')
plt.title('Fare Distribution by Passenger Class and Gender')
plt.xlabel('Passenger Class')
plt.ylabel('Fare')
plt.legend(title='Sex')
plt.show()
```

6-SCATTER PLOT

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=data, x='Age', y='Fare', hue='Survived')
plt.title('Relationship between Age and Fare by Survival Status')
plt.xlabel('Age')
plt.ylabel('Fare')
plt.legend(title='Survived', labels=['No','Yes'])
plt.show()
```
